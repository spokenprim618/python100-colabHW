## Task 1

```python
ecommerce = pd.read_csv("/root/.cache/kagglehub/datasets/salahuddinahmedshuvo/ecommerce-consumer-behavior-analysis-data/versions/1/Ecommerce_Consumer_Behavior_Analysis_Data.csv")
print(ecommerce.head(5))
def fix_purchase(purchase):
    new_string =""
    for i in range(1,len(purchase)):
        new_string+=purchase[i]
    return float(new_string)
ecommerce['Purchase_Amount'] = ecommerce['Purchase_Amount'].map(fix_purchase)
print(ecommerce['Purchase_Amount'].sum())


buying_patterns = pd.pivot_table(ecommerce,index=['Purchase_Category'], columns=['Gender','Income_Level'],values=['Purchase_Amount'],aggfunc = 'sum', fill_value=0)
print(buying_patterns)

demographics = pd.pivot_table(ecommerce,index=['Income_Level','Education_Level'], columns=['Marital_Status'],values=['Customer_ID'],aggfunc = 'count', fill_value=0)
print(demographics)
```

## Task 2

```python
jobs = pd.read_csv('/kaggle/input/ai-powered-job-recommendations/job_recommendation_dataset.csv')
print(jobs.head(10))

def change(row):
    if row['Experience Level'] == 'Entry Level' and row['Salary'] > 70000:
        if 'Python' in  row['Required Skills']:
            if 'SQL' in  row['Required Skills']:
                return 'yes'
    else:
        return 'no'
jobs['Check These Out'] = jobs.apply(change,axis=1)
print(jobs)
my_jobs = jobs[jobs['Check These Out']=='yes']
print(my_jobs)
```
